```python
import numpy as np
import pandas as pd


def generate_and_save_random_data(shape, filename):
    """
    Generates random float data with the given 2D shape and saves it to a CSV file.

    Parameters:
      shape (tuple): A tuple of two integers (rows, cols).
      filename (str): The name of the CSV file to save the data.
    """

    data = np.random.rand(*shape)
    df = pd.DataFrame(data)
    df.to_csv(filename, index=False, header=False)


channel_dim = [56, 28, 14, 7]
channel_sizes = [c * c for c in channel_dim]


for i, size in enumerate(channel_sizes):
    shape = (1, size)
    generate_and_save_random_data(shape, f"1_{size}.csv")
```
